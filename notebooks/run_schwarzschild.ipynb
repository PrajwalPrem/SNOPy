# S2 orbit fit for Schwarzschild

Full pipeline for the Schwarzschild case. All the physics and priors are in, `snope/metrics/schwarzschild.py`. Changes should be made there. There's no new-physics parameter here. The gravitational parameters is just the Mass of the supermassive black hole.

````python
from snope.metrics.schwarzschild import main, metric, param_priors
from snope.priors import PriorMode
from snope.orbit_model import S2OrbitModel
from snope import plotting
````

## Effective potential

We need to check if the effective potential `V_eff(r)` actually has two turning points (periapsis and apoapsis) at some parameter values. If this requirement is not met, the integrator will fail. Try a few different values of `a`/`e` to get it right.

````python
model = S2OrbitModel(metric, "../data/tab_gillessen_pos.csv", "../data/tab_gillessen_vr.csv", verbose=False)

preview = model.preview_effective_potential(
    M_bh=4.3e6, distance=8.33, a=125.5, e=0.884,
)
plotting.effective_potential_plot(preview, "Schwarzschild")
````

## Run the mcmc

By default we use mixed priors: flat priors for orbital elements, Gaussian for the offsets and `t_peri`.

````python
sampler, flat_samples, best_fit = main(
    data_path_pos="../data/tab_gillessen_pos.csv",
    data_path_rv="../data/tab_gillessen_vr.csv",
)
best_fit
````

## Other settings examples

```python
# flat prior for every parameter
main(prior_mode=PriorMode.FLAT)

# Gaussian prior for every parameter 
main(prior_mode=PriorMode.GAUSSIAN)

# override any pipeline setting (n_steps, n_walkers, n_points, ...)
main(prior_mode=PriorMode.MIXED, n_walkers=64, n_steps=50000, burn_in=5000)

# a quick run to check if it works 
main(n_walkers=8, n_steps=200, burn_in=20, n_points=200)
```
