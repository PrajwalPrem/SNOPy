# S2 orbit fit for Schwarzschild-de Sitter

This notebook runs the full pipeline for Schwarzschild de Sitter (SdS) metric with g_tt = 1/g_rr = 1 - 2M/r - (Lambda_geo * r^2)/3

All the physics and priors live in one file, `snope/metrics/sds.py`. Corrections can be made there.

````python
from snope.metrics.sds import main, metric, param_priors
from snope.priors import PriorMode
from snope.orbit_model import S2OrbitModel
from snope import plotting
````

## Effective potential

Before a full MCMC run, we check if the effective potential `V_eff(r)` actually has two turning points (periapsis and apoapsis) at some trial parameter values. It is important to check this. Try a few different values of the new parameter here (and of `a`/`e`).

````python
model = S2OrbitModel(metric, "../data/tab_gillessen_pos.csv", "../data/tab_gillessen_vr.csv", verbose=False)

preview = model.preview_effective_potential(
    M_bh=4.3e6, distance=8.33, a=125.5, e=0.884,
    log10_lambda=-52.0,
)
plotting.effective_potential_plot(preview, "Schwarzschild-de Sitter")
````

## Run the mcmc

By default we use mixed priors. A flat prior is taken on new-physics parameter (`log10_lambda`).

````python
sampler, flat_samples, best_fit = main(
    data_path_pos="../data/tab_gillessen_pos.csv",
    data_path_rv="../data/tab_gillessen_vr.csv",
)
best_fit
````

## Some settings examples

```python
# every parameter has flat prior
main(prior_mode=PriorMode.FLAT)

# every parameter has Gaussian prior
main(prior_mode=PriorMode.GAUSSIAN)

# override any setting (n_steps, n_walkers, n_points, ...)
main(prior_mode=PriorMode.MIXED, n_walkers=64, n_steps=50000, burn_in=5000)

# just a quick run if you want to check if it works or not
main(n_walkers=8, n_steps=200, burn_in=20, n_points=200)
```
